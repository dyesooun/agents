asyncio - used by all agents
- lightweight of multi-threading
- does not involve threads/multi processing

```
async def do_some_processing() -> str:
    # Do something
    return "done";

result = await do_some_processing()
```

functions defined with `async def` are called coroutines -> special functions that can be paused and resumed

calling a coroutine doesn't execute immediately -> it returns a coroutine object instead

to run,
- you must `await` it -> which schedules it for execution within an event loop

while a coroutine is waiting (e.g. for IO), the event loop can run other coroutines. Basically if the current one is paused (waiting for something), the next corouting will run. Not really as multi-threading per se.
Multi-threading, not on OS level but like brute forcing it.

to run multiple coroutines and schedule all of them, use:
```
results = await asyncio.gather(
    do_something(),
    do_something_else(),
    do_something_more()
)
```

To run an agents:
1. create an instance of agent
2. use `with trace()` to track the agent (Optional)
3. call `runner.run()` to run the agent -> a coroutine

````python
from dotenv import load_dotenv
from agents import Agent, Runner, trace
````

````python
load_dotenv(override=True)
````

````python
#creating an agent

agent = Agent(
    name="Jokester",
    instructions="You are a joke teller", #the system_prompt
    model="gpt-4o-mini"
)
````

````python
with trace("Telling a joke"):
    results = await Runner.run(agent, "Tell a joke about AI in an apocalyptic world brought about by rogue AI")
    print(results.final_output)
````
